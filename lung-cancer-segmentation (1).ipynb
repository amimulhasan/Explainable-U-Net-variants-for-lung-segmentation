```python
# =====================================================================================
# EXPLAINABLE U-NET VARIANTS FOR LUNG SEGMENTATION ON CHEST X-RAYS
# Based on: Abushahla & Arslan (2026), Tuberk Toraks 74(2):84-97
# Implements: Baseline U-Net | Attention U-Net | Shallow U-Net
# Pipeline: Dataset -> Preprocessing -> K-Fold Training -> Evaluation -> Grad-CAM XAI
#
# EI SCRIPT KAGGLE NOTEBOOK E CELL BY CELL COPY-PASTE KORE RUN KORO.
# Each "# %% [CELL N]" ekta notebook cell hisebe treat koro.
# =====================================================================================

# %% [CELL 1] -------------------- IMPORTS --------------------
import os
import re
import glob
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2

import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, backend as K
from sklearn.model_selection import KFold

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))


# %% [CELL 2] -------------------- PATHS (EDIT THESE) --------------------
# Kaggle e tomar dataset shadharonoto eikhane thake:
#   /kaggle/input/<dataset-folder-name>/images/
#   /kaggle/input/<dataset-folder-name>/masks/
#
# Niche IMAGE_DIR ar MASK_DIR change kore tomar actual path disho.
# Kaggle right-side panel e dataset add korar por path dekhte parbe.

IMAGE_DIR = os.environ.get("IMAGE_DIR", "data/images")
# Kaggle fallback:
if not os.path.isdir(IMAGE_DIR):
    IMAGE_DIR = "/kaggle/input/datasets/amimulahasanrofik/chest-x-ray-segmentation/image"
MASK_DIR = os.environ.get("MASK_DIR", "data/masks")
if not os.path.isdir(MASK_DIR):
    MASK_DIR = "/kaggle/input/datasets/amimulahasanrofik/chest-x-ray-segmentation/Masks"

IMG_SIZE = 256          # paper-er moto 256x256
BATCH_SIZE = 8
EPOCHS = 50            # paper e 100, tumi compute er upor base kore kom dite paro (e.g. 30-50)
N_FOLDS = 5             # paper e k=5, fast test korte chaile 2-3 dite paro
OUTPUT_DIR = os.environ.get("OUTPUT_DIR", "outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)


# %% [CELL 3] -------------------- MATCH IMAGE-MASK PAIRS --------------------
# Paper-er moto: filename match kore image-mask pair banano, jegula match hoyna
# segula baad dewa (one-to-one correspondence).

def list_files(folder, exts=(".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff")):
    files = []
    for ext in exts:
        files.extend(glob.glob(os.path.join(folder, f"*{ext}")))
        files.extend(glob.glob(os.path.join(folder, f"*{ext.upper()}")))
    return sorted(files)

def get_stem(path):
    """Filename extension chhara base name return kore, mask suffix thakle clean kore."""
    name = os.path.splitext(os.path.basename(path))[0]
    # common mask suffix patterns clean up (dorkar na hole eta affect korbe na)
    name = re.sub(r"(_mask|_masks|-mask|mask_)$", "", name, flags=re.IGNORECASE)
    return name

image_files = list_files(IMAGE_DIR)
mask_files  = list_files(MASK_DIR)

print(f"Found {len(image_files)} images, {len(mask_files)} masks (before matching)")

image_dict = {get_stem(p): p for p in image_files}
mask_dict  = {get_stem(p): p for p in mask_files}

common_stems = sorted(set(image_dict.keys()) & set(mask_dict.keys()))
paired_images = [image_dict[s] for s in common_stems]
paired_masks  = [mask_dict[s] for s in common_stems]

print(f"Matched image-mask pairs: {len(paired_images)}")
assert len(paired_images) > 0, "Kono pair match korenai! IMAGE_DIR / MASK_DIR path check koro."


# %% [CELL 4] -------------------- EDA: QUICK VISUAL CHECK --------------------
# Random kichu image-mask overlay dekhe niye check kora je sob thik ache.

def quick_eda(n=4):
    idxs = random.sample(range(len(paired_images)), min(n, len(paired_images)))
    fig, axes = plt.subplots(2, n, figsize=(4*n, 8))
    for i, idx in enumerate(idxs):
        img = cv2.imread(paired_images[idx], cv2.IMREAD_GRAYSCALE)
        msk = cv2.imread(paired_masks[idx], cv2.IMREAD_GRAYSCALE)
        axes[0, i].imshow(img, cmap="gray")
        axes[0, i].set_title("Image")
        axes[0, i].axis("off")
        axes[1, i].imshow(img, cmap="gray")
        axes[1, i].imshow(msk, cmap="jet", alpha=0.4)
        axes[1, i].set_title("Mask overlay")
        axes[1, i].axis("off")
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "eda_samples.png"), dpi=120)
    plt.show()

quick_eda()


# %% [CELL 5] -------------------- PREPROCESSING --------------------
# Paper-er moto: resize 256x256, image normalize [0,1], mask binarize at 0.5,
# 80:20 train/val split (kintu amra cross-validation o korbo).

def load_image(path, size=IMG_SIZE):
    img = cv2.imread(path, cv2.IMREAD_COLOR)          # 3-channel input (paper input: 256,256,3)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (size, size), interpolation=cv2.INTER_AREA)
    img = img.astype(np.float32) / 255.0
    return img

def load_mask(path, size=IMG_SIZE):
    msk = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    msk = cv2.resize(msk, (size, size), interpolation=cv2.INTER_NEAREST)
    msk = msk.astype(np.float32) / 255.0
    msk = (msk > 0.5).astype(np.float32)              # binarize at 0.5
    msk = np.expand_dims(msk, axis=-1)                 # (H, W, 1)
    return msk

def load_dataset(image_paths, mask_paths, size=IMG_SIZE):
    X = np.zeros((len(image_paths), size, size, 3), dtype=np.float32)
    Y = np.zeros((len(mask_paths), size, size, 1), dtype=np.float32)
    for i, (ip, mp) in enumerate(zip(image_paths, mask_paths)):
        X[i] = load_image(ip, size)
        Y[i] = load_mask(mp, size)
    return X, Y

print("Loading full dataset into memory (eta kichu shomoy nibe)...")
X_all, Y_all = load_dataset(paired_images, paired_masks)
print("X_all:", X_all.shape, "Y_all:", Y_all.shape)

# NOTE: dataset boro hole (e.g. 1000+ images at 256x256x3) shob memory te load
# na kore tf.data.Dataset / generator use kora better. Choto-medium dataset
# (paper er moto ~700 image) er jonno eta thik ache.


# %% [CELL 6] -------------------- METRICS: IoU & DICE --------------------
# Paper-er formula:  IoU = TP / (TP+FP+FN)   Dice = 2TP / (2TP+FP+FN)

def dice_coef(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def dice_loss(y_true, y_pred):
    return 1 - dice_coef(y_true, y_pred)

def iou_metric(y_true, y_pred, smooth=1e-6):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.cast(K.flatten(y_pred) > 0.5, tf.float32)
    intersection = K.sum(y_true_f * y_pred_f)
    union = K.sum(y_true_f) + K.sum(y_pred_f) - intersection
    return (intersection + smooth) / (union + smooth)

def bce_loss(y_true, y_pred):
    return tf.keras.losses.binary_crossentropy(y_true, y_pred)


# %% [CELL 7] -------------------- MODEL 1: BASELINE U-NET --------------------
# Figure 2 onujayi: 3-block encoder (64,128,256 filters) -> bottleneck (512)
# -> 3-block decoder with skip connections -> sigmoid output

def conv_block(x, filters, name_prefix):
    x = layers.Conv2D(filters, 3, padding="same", activation="relu", name=f"{name_prefix}_conv1")(x)
    x = layers.Conv2D(filters, 3, padding="same", activation="relu", name=f"{name_prefix}_conv2")(x)
    return x

def build_unet(input_size=(IMG_SIZE, IMG_SIZE, 3)):
    inputs = layers.Input(shape=input_size, name="input")

    # Encoder
    c1 = conv_block(inputs, 64, "enc1")
    p1 = layers.MaxPooling2D(2, name="pool1")(c1)

    c2 = conv_block(p1, 128, "enc2")
    p2 = layers.MaxPooling2D(2, name="pool2")(c2)

    c3 = conv_block(p2, 256, "enc3")
    p3 = layers.MaxPooling2D(2, name="pool3")(c3)

    # Bottleneck
    bn = conv_block(p3, 512, "bottleneck")

    # Decoder
    u4 = layers.UpSampling2D(2, name="up4")(bn)
    u4 = layers.Concatenate(name="concat4")([u4, c3])
    d4 = conv_block(u4, 256, "dec4")

    u5 = layers.UpSampling2D(2, name="up5")(d4)
    u5 = layers.Concatenate(name="concat5")([u5, c2])
    d5 = conv_block(u5, 128, "dec5")

    u6 = layers.UpSampling2D(2, name="up6")(d5)
    u6 = layers.Concatenate(name="concat6")([u6, c1])
    d6 = conv_block(u6, 64, "dec6")

    outputs = layers.Conv2D(1, 1, activation="sigmoid", name="output_conv")(d6)

    model = models.Model(inputs, outputs, name="Baseline_UNet")
    return model


# %% [CELL 8] -------------------- MODEL 2: ATTENTION U-NET --------------------
# Figure 3 onujayi: same encoder/decoder, kintu skip connection ki Attention
# Gate diye filter kora hoy ag-e concatenate korar age.

def attention_gate(x, g, inter_channels, name_prefix):
    """
    x: encoder feature (skip connection)
    g: gating signal from decoder
    """
    theta_x = layers.Conv2D(inter_channels, 1, padding="same", name=f"{name_prefix}_theta_x")(x)
    phi_g   = layers.Conv2D(inter_channels, 1, padding="same", name=f"{name_prefix}_phi_g")(g)

    # shapes match na hole resize kora (encoder feature map decoder gate er chaiye boro hote pare)
    if theta_x.shape[1] != phi_g.shape[1]:
        phi_g = layers.UpSampling2D(
            size=(theta_x.shape[1] // phi_g.shape[1], theta_x.shape[2] // phi_g.shape[2]),
            name=f"{name_prefix}_upsample_g"
        )(phi_g)

    add_xg = layers.Add(name=f"{name_prefix}_add")([theta_x, phi_g])
    relu_xg = layers.Activation("relu", name=f"{name_prefix}_relu")(add_xg)
    psi = layers.Conv2D(1, 1, padding="same", name=f"{name_prefix}_psi")(relu_xg)
    alpha = layers.Activation("sigmoid", name=f"{name_prefix}_sigmoid")(psi)

    attended = layers.Multiply(name=f"{name_prefix}_multiply")([x, alpha])
    return attended

def build_attention_unet(input_size=(IMG_SIZE, IMG_SIZE, 3)):
    inputs = layers.Input(shape=input_size, name="input")

    # Encoder (same as baseline)
    c1 = conv_block(inputs, 64, "enc1")
    p1 = layers.MaxPooling2D(2, name="pool1")(c1)

    c2 = conv_block(p1, 128, "enc2")
    p2 = layers.MaxPooling2D(2, name="pool2")(c2)

    c3 = conv_block(p2, 256, "enc3")
    p3 = layers.MaxPooling2D(2, name="pool3")(c3)

    # Bottleneck
    bn = conv_block(p3, 512, "bottleneck")

    # Decoder with Attention Gates on each skip connection
    g4 = layers.UpSampling2D(2, name="gate_up4")(bn)
    att3 = attention_gate(c3, g4, inter_channels=256, name_prefix="ag3")
    u4 = layers.UpSampling2D(2, name="up4")(bn)
    u4 = layers.Concatenate(name="concat4")([u4, att3])
    d4 = conv_block(u4, 256, "dec4")

    g5 = layers.UpSampling2D(2, name="gate_up5")(d4)
    att2 = attention_gate(c2, g5, inter_channels=128, name_prefix="ag2")
    u5 = layers.UpSampling2D(2, name="up5")(d4)
    u5 = layers.Concatenate(name="concat5")([u5, att2])
    d5 = conv_block(u5, 128, "dec5")

    g6 = layers.UpSampling2D(2, name="gate_up6")(d5)
    att1 = attention_gate(c1, g6, inter_channels=64, name_prefix="ag1")
    u6 = layers.UpSampling2D(2, name="up6")(d5)
    u6 = layers.Concatenate(name="concat6")([u6, att1])
    d6 = conv_block(u6, 64, "dec6")

    outputs = layers.Conv2D(1, 1, activation="sigmoid", name="output_conv")(d6)

    model = models.Model(inputs, outputs, name="Attention_UNet")
    return model


# %% [CELL 9] -------------------- MODEL 3: SHALLOW U-NET --------------------
# Figure 4 onujayi: 2-block encoder (32, 64 filters) -> bottleneck (128)
# -> 2-block decoder -> sigmoid output. Lightweight, fast inference.
#
# NOTE: Baseline/Attention U-Net e protekti block e DOUBLE conv (Fig 2,3),
# kintu Shallow U-Net e Fig 4 onujayi protekti block e shudhu EKTA conv layer
# (eta i lightweight design-er main reason). Ei single_conv_block use kora
# hoyeche jate param count paper-er 231,617 er sathe exact match kore.

def single_conv_block(x, filters, name_prefix):
    x = layers.Conv2D(filters, 3, padding="same", activation="relu", name=f"{name_prefix}_conv1")(x)
    return x

def build_shallow_unet(input_size=(IMG_SIZE, IMG_SIZE, 3)):
    inputs = layers.Input(shape=input_size, name="input")

    # Encoder (2 blocks only, single conv per block)
    c1 = single_conv_block(inputs, 32, "enc1")
    p1 = layers.MaxPooling2D(2, name="pool1")(c1)

    c2 = single_conv_block(p1, 64, "enc2")
    p2 = layers.MaxPooling2D(2, name="pool2")(c2)

    # Bottleneck
    bn = single_conv_block(p2, 128, "bottleneck")

    # Decoder (2 blocks only, single conv per block)
    u3 = layers.UpSampling2D(2, name="up3")(bn)
    u3 = layers.Concatenate(name="concat3")([u3, c2])
    d3 = single_conv_block(u3, 64, "dec3")

    u4 = layers.UpSampling2D(2, name="up4")(d3)
    u4 = layers.Concatenate(name="concat4")([u4, c1])
    d4 = single_conv_block(u4, 32, "dec4")

    outputs = layers.Conv2D(1, 1, activation="sigmoid", name="output_conv")(d4)

    model = models.Model(inputs, outputs, name="Shallow_UNet")
    return model


# %% [CELL 10] -------------------- SANITY CHECK: PARAM COUNTS --------------------
# Paper-er Table 1 e: U-Net ~7.78M, Attention U-Net ~8.04M, Shallow U-Net ~0.23M
# Tomar version e exact match na hote pare (architecture details slightly vary
# korte pare), kintu order of magnitude similar thakbe.

for name, builder in [("U-Net", build_unet),
                       ("Attention U-Net", build_attention_unet),
                       ("Shallow U-Net", build_shallow_unet)]:
    m = builder()
    n_params = m.count_params()
    print(f"{name}: {n_params:,} parameters")
    del m
    K.clear_session()


# %% [CELL 11] -------------------- TRAINING FUNCTION (K-FOLD) --------------------
# Paper-er Table 1: Adam optimizer, binary cross-entropy loss, 100 epochs,
# batch size 8, 5-fold cross-validation.

def compile_model(model, lr=1e-4):
    model.compile(
        optimizer=optimizers.Adam(learning_rate=lr),
        loss=bce_loss,
        metrics=[dice_coef, iou_metric, "accuracy"]
    )
    return model

def train_with_kfold(builder_fn, model_name, X, Y, n_folds=N_FOLDS,
                      epochs=EPOCHS, batch_size=BATCH_SIZE):
    kf = KFold(n_splits=n_folds, shuffle=True, random_state=SEED)
    fold_histories = []
    fold_metrics = []
    best_model = None
    best_dice = -1

    for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
        print(f"\n===== {model_name} | Fold {fold+1}/{n_folds} =====")
        X_train, X_val = X[train_idx], X[val_idx]
        Y_train, Y_val = Y[train_idx], Y[val_idx]

        K.clear_session()
        model = compile_model(builder_fn())

        early_stop = tf.keras.callbacks.EarlyStopping(
            monitor="val_loss", patience=10, restore_best_weights=True
        )
        reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss", factor=0.5, patience=5, min_lr=1e-7
        )

        history = model.fit(
            X_train, Y_train,
            validation_data=(X_val, Y_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=[early_stop, reduce_lr],
            verbose=1
        )

        val_loss, val_dice, val_iou, val_acc = model.evaluate(X_val, Y_val, verbose=0)
        print(f"Fold {fold+1} -> Dice: {val_dice:.4f} | IoU: {val_iou:.4f} | Acc: {val_acc:.4f}")

        fold_histories.append(history.history)
        fold_metrics.append({"fold": fold+1, "dice": val_dice, "iou": val_iou, "accuracy": val_acc})

        if val_dice > best_dice:
            best_dice = val_dice
            best_model = model

    metrics_df = pd.DataFrame(fold_metrics)
    print(f"\n{model_name} | Mean across folds:")
    print(metrics_df.mean(numeric_only=True))

    return best_model, metrics_df, fold_histories


# %% [CELL 12] -------------------- RUN TRAINING FOR ALL 3 MODELS --------------------
# WARNING: ei cell e 5-fold x 100-epoch x 3-model = onek shomoy nibe.
# Test korar jonno EPOCHS = 10-20 ar N_FOLDS = 2-3 kore age try koro,
# tarpor full run dio.

import time

results = {}
trained_models = {}

model_builders = {
    "U-Net": build_unet,
    "Attention U-Net": build_attention_unet,
    "Shallow U-Net": build_shallow_unet,
}

for name, builder in model_builders.items():
    start = time.time()
    best_model, metrics_df, histories = train_with_kfold(builder, name, X_all, Y_all)
    elapsed = time.time() - start

    results[name] = {
        "metrics_df": metrics_df,
        "mean_dice": metrics_df["dice"].mean(),
        "mean_iou": metrics_df["iou"].mean(),
        "mean_accuracy": metrics_df["accuracy"].mean(),
        "runtime_sec": elapsed,
    }
    trained_models[name] = best_model

    # save best model of each architecture
    best_model.save(os.path.join(OUTPUT_DIR, f"{name.replace(' ', '_')}.keras"))
    print(f"{name} total training time: {elapsed:.1f}s")


# %% [CELL 13] -------------------- SUMMARY TABLE (LIKE PAPER'S TABLE 2) --------------------

summary_rows = []
for name, res in results.items():
    summary_rows.append({
        "Model": name,
        "Accuracy": round(res["mean_accuracy"], 4),
        "IoU": round(res["mean_iou"], 4),
        "Dice": round(res["mean_dice"], 4),
        "Runtime (s)": round(res["runtime_sec"], 1),
    })

summary_df = pd.DataFrame(summary_rows)
print(summary_df)
summary_df.to_csv(os.path.join(OUTPUT_DIR, "model_comparison_summary.csv"), index=False)


# %% [CELL 14] -------------------- QUALITATIVE COMPARISON: PREDICTIONS VS GROUND TRUTH --------------------

def show_predictions(models_dict, X, Y, n_samples=3):
    idxs = random.sample(range(len(X)), n_samples)
    n_models = len(models_dict)
    fig, axes = plt.subplots(n_samples, n_models + 2, figsize=(3*(n_models+2), 3*n_samples))

    for row, idx in enumerate(idxs):
        img = X[idx]
        gt = Y[idx, :, :, 0]

        axes[row, 0].imshow(img)
        axes[row, 0].set_title("Original")
        axes[row, 0].axis("off")

        axes[row, 1].imshow(img)
        axes[row, 1].imshow(gt, cmap="jet", alpha=0.4)
        axes[row, 1].set_title("Ground Truth")
        axes[row, 1].axis("off")

        for col, (name, model) in enumerate(models_dict.items()):
            pred = model.predict(img[np.newaxis, ...], verbose=0)[0, :, :, 0]
            pred_bin = (pred > 0.5).astype(np.float32)
            axes[row, col+2].imshow(img)
            axes[row, col+2].imshow(pred_bin, cmap="jet", alpha=0.4)
            axes[row, col+2].set_title(name)
            axes[row, col+2].axis("off")

    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "qualitative_comparison.png"), dpi=120)
    plt.show()

show_predictions(trained_models, X_all, Y_all, n_samples=3)


# %% [CELL 15] -------------------- GRAD-CAM IMPLEMENTATION --------------------
# Paper-er moto: final convolutional layer e Grad-CAM apply kore dekha hoy
# model lung region e focus korche kina.

def get_last_conv_layer_name(model):
    for layer in reversed(model.layers):
        if isinstance(layer, layers.Conv2D) and layer.name != "output_conv":
            return layer.name
    raise ValueError("Kono Conv2D layer pawa jay nai.")

def make_gradcam_heatmap(model, img_array, layer_name=None):
    if layer_name is None:
        layer_name = get_last_conv_layer_name(model)

    grad_model = models.Model(
        inputs=model.inputs,
        outputs=[model.get_layer(layer_name).output, model.output]
    )

    input_tensor = tf.convert_to_tensor(img_array[np.newaxis, ...], dtype=tf.float32)
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(input_tensor, training=False)
        # Segmentation task: "class score" = mean predicted mask activation
        loss = tf.reduce_mean(predictions)

    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]
    heatmap = tf.reduce_sum(tf.multiply(pooled_grads, conv_outputs), axis=-1)
    heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)
    heatmap = heatmap.numpy()
    heatmap = cv2.resize(heatmap, (img_array.shape[1], img_array.shape[0]))
    return heatmap

def show_gradcam_comparison(models_dict, X, Y, n_samples=3):
    idxs = random.sample(range(len(X)), n_samples)
    n_models = len(models_dict)
    fig, axes = plt.subplots(n_samples, n_models + 1, figsize=(3.5*(n_models+1), 3.5*n_samples))

    for row, idx in enumerate(idxs):
        img = X[idx]
        axes[row, 0].imshow(img)
        axes[row, 0].set_title("Original")
        axes[row, 0].axis("off")

        for col, (name, model) in enumerate(models_dict.items()):
            heatmap = make_gradcam_heatmap(model, img)
            axes[row, col+1].imshow(img)
            im = axes[row, col+1].imshow(heatmap, cmap="jet", alpha=0.5)
            axes[row, col+1].set_title(f"{name} Grad-CAM")
            axes[row, col+1].axis("off")

    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, "gradcam_comparison.png"), dpi=120)
    plt.show()

show_gradcam_comparison(trained_models, X_all, Y_all, n_samples=3)


# %% [CELL 16] -------------------- PER-SAMPLE METRICS (OPTIONAL: FOR FIGURE 5/6 STYLE) --------------------
# Individual image-er IoU/Dice/F1 dekhar jonno, jate tumi best/worst case
# samples ber kore inspect korte paro (paper-er Figure 5 & 6 er moto).

def compute_sample_metrics(model, X, Y):
    preds = model.predict(X, batch_size=BATCH_SIZE, verbose=0)
    preds_bin = (preds > 0.5).astype(np.float32)

    records = []
    for i in range(len(X)):
        y_true = Y[i, :, :, 0].flatten()
        y_pred = preds_bin[i, :, :, 0].flatten()
        tp = np.sum((y_true == 1) & (y_pred == 1))
        fp = np.sum((y_true == 0) & (y_pred == 1))
        fn = np.sum((y_true == 1) & (y_pred == 0))
        iou = tp / (tp + fp + fn + 1e-8)
        dice = (2*tp) / (2*tp + fp + fn + 1e-8)
        records.append({"index": i, "iou": iou, "dice": dice})
    return pd.DataFrame(records)

for name, model in trained_models.items():
    df = compute_sample_metrics(model, X_all, Y_all)
    df.to_csv(os.path.join(OUTPUT_DIR, f"per_sample_metrics_{name.replace(' ', '_')}.csv"), index=False)
    print(f"\n{name}: worst 3 samples by IoU")
    print(df.sort_values("iou").head(3))


# %% [CELL 17] -------------------- DONE --------------------
print("\nAll done! Outputs saved in:", OUTPUT_DIR)
print(" - model_comparison_summary.csv  (Table 2 style)")
print(" - qualitative_comparison.png")
print(" - gradcam_comparison.png")
print(" - per_sample_metrics_<model>.csv")
print(" - <model>.keras  (trained model weights)")
```
